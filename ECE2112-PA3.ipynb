## A. POSITIONAL AND LABEL-BASED SLICING

Load the cars.csv dataset, then extract rows 6–10 and display selected columns using label-based indexing.

```python
# Load the cars CSV file into a DataFrame 
cars = pd.read_csv("cars.csv")

print("a.)")

#Display the shape of cars 
print("Shape of cars:", cars.shape)

# Display the complete list of column names of cars
print("Column names:", list(cars.columns))

print("\nb.)")

# Use iloc for positional slicing to get rows 6-10, where the first data row is row 1.
cars_6_to_10 = cars.iloc[5:10]
print("cars_6_to_10:\n", cars_6_to_10)

print("\nc.)")

#Display only Model, mpg, cyl, hp, and gear, in that order
Display_Columns = cars_6_to_10[['Model', 'mpg', 'cyl', 'hp', 'gear']]
print("Display Columns:\n", Display_Columns)
```

## B. MODEL LOOKUP

Use Boolean indexing on the model column to retrieve the full record for Toyota Corolla and selected columns for Pontiac Firebird.

```python
print("a.)")

# Use a Boolean condition on the Model column to find Toyota Corolla
toyota = cars[cars['Model']=='Toyota Corolla']
print("toyota:\n", toyota)

print("\nb.)")

# Use a Boolean condition to find Pontiac Firebird, then keep only the selected columns
pontiac = cars[cars['Model']=='Pontiac Firebird'][['Model', 'mpg', 'hp', 'wt']]
print("pontiac:\n", pontiac)
```

## C. MULTI-MODEL SUBSETTING

 Select the records for Datsun 710, Lotus Europa, and Ferrari Dino by their model names, keeping only Model, mpg, cyl, hp, and gear.

```python
# Use Boolean OR conditions to select the rows by their model values rather than by row numbers; keep only the required columns.
selected_cars = cars.loc[(cars['Model']=='Datsun 710')|(cars['Model']=='Lotus Europa')|(cars['Model']=='Ferrari Dino'), ['Model', 'mpg', 'cyl', 'hp', 'gear']]
print("selected_cars:\n", selected_cars)
print("\nShape of selected_cars:", selected_cars.shape)
```
